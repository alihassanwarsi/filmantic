## Observation
- `genres` is plain text that looks like a list but its not a real list
- `keywords` has the same problem
- `cast` is plain text too with lots of extra info (character, gender...). We only need the name
- `crew` has many people (editor, designer...). We only need the one whose job is "Director"

```python
import ast
```

```python
def convert(text):
    try:
        data = ast.literal_eval(text)
        return [item["name"] for item in data]
    except (ValueError, TypeError, SyntaxError):
        return []
```

```python
movies_df["genres"] = movies_df["genres"].apply(convert)
movies_df["keywords"] = movies_df["keywords"].apply(convert)

movies_df[["title", "genres", "keywords"]].head(5)
```

```python
movies_df["cast"] = movies_df["cast"].apply(lambda t: convert(t)[:3])

movies_df[["title", "cast"]].head(5)
```

```python
def get_director(text):
    for item in ast.literal_eval(text):
        if item["job"] == "Director":
            return item["name"]
    return ""
```

```python
movies_df["director"] = movies_df["crew"].apply(get_director)

movies_df[["title", "director"]].head(5)
```

```python
final_df = movies_df.drop(columns="crew")
print(final_df.shape)
```

## Final data
Genres, keywords, cast and crew came as plain text.
I turned them into proper lists, kept the top 3 actors, and picked the director out of the crew.

```python
import pandas as pd

movies = pd.read_csv("../data/raw/tmdb_5000_movies.csv")
credits = pd.read_csv("../data/raw/tmdb_5000_credits.csv")
```

```python
by_title = movies.merge(credits, on="title")
by_id = movies.merge(credits.drop(columns="title"), left_on="id", right_on="movie_id")

print("original:", movies.shape[0])
print("merge on title:", by_title.shape[0],)
print("merge on id:", by_id.shape[0])
```

```python
wrong = by_title[by_title["id"] != by_title["movie_id"]]
print(len(wrong))
wrong[["id", "movie_id", "title"]].head(10)
```

Merging on `title` gives 4809 rows instead of 4803 and swaps cast/crew for 3 titles. While merging on `movie_id` gives the correct 4803.

```python
by_id.isnull().sum()
```

```python
movies_df = by_id[["movie_id", "title", "overview", "genres", "keywords", "cast", "crew"]]
```

```python
movies_df = movies_df.dropna(subset=["overview"]).reset_index(drop=True)
print(movies_df.shape)
```

## Findings
- Merge on movie id, not title.
- 3 movies have no overview: drop them and reset the index
- `tagline` is missing for 844 movies, so we don't use it.
